```python
# Program 4: Gradient Descent Algorithm to Find Local Minima
# Objective Function: y = (x + 3)^2, Starting x = 2, Learning Rate = 0.1

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

def function(x):
    return (x + 3) ** 2

def gradient(x):
    return 2 * (x + 3)
```

```python
# Gradient Descent Function
def gradient_descent(starting_point=2.0, learning_rate=0.1, num_iterations=100, tolerance=1e-8):
    x = starting_point
    x_history = [x]
    y_history = [function(x)]

    for i in range(1, num_iterations + 1):
        grad = gradient(x)
        x_new = x - learning_rate * grad
        x_history.append(x_new)
        y_history.append(function(x_new))

        if abs(grad) < tolerance:
            print(f"Converged at iteration {i}!")
            break
        x = x_new

    return x_history, y_history

x_hist, y_hist = gradient_descent(starting_point=2.0, learning_rate=0.1, num_iterations=100)
print(f"Local Minima found at: x = {x_hist[-1]:.6f}, y = {y_hist[-1]:.6f}")
print(f"Expected Minimum:      x = -3.000000, y = 0.000000")
```

```python
# Visualization
x_plot = np.linspace(-8, 5, 200)
y_plot = function(x_plot)

plt.figure(figsize=(10, 6))
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='Cost Function $y=(x+3)^2$')
plt.plot(x_hist, y_hist, 'ro-', markersize=4, linewidth=1, label='Gradient Descent Steps', alpha=0.7)
plt.plot(x_hist[0], y_hist[0], 'gs', markersize=12, label=f'Start (x={x_hist[0]})')
plt.plot(x_hist[-1], y_hist[-1], 'r*', markersize=15, label=f'Minimum (x={x_hist[-1]:.2f})')
plt.title('Gradient Descent: Finding Local Minima of $y = (x + 3)^2$')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
